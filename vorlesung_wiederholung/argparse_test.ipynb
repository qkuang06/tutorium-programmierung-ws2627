```python
import argparse

parser = argparse.ArgumentParser(description='der parser ist fürs skript argparse_test') # eine beschreibung worauf man bei der angabe des arguments achten soll
```

```python
# das argument ohne bindestrich ist positional, also beim aufrufen verpflichtend anzugeben
parser.add_argument('file', help='eine datei muss gegeben werden')
```

```python
# das argument mit einem bindestrich ist optional wie -n und -q
parser.add_argument('-n', '--number', type=int, default=1, help='eine zahl ist bei bedarf anzugeben, die ist kein muss')
```

```python
parser.add_argument('-q', '--quiet', action='store_true', default=False, help='das argument ist auch erst bei bedarf anzugeben')
# Normalerweise erwartet ein optionales Argument einen Wert dahinter z.B. -n 5
# mit action='store_true' muss man den boolean Wert nicht eintippen, sondern den durch bloßes hinschreiben aktivieren
# -q statt -q True 
```

```python
args = parser.parse_args() 
# das ist der moment des ausführens, das argument-parser liest die Angaben aus der Kommandozeile aus
```

```python
print(args.file, args.number, args.quiet)
```
