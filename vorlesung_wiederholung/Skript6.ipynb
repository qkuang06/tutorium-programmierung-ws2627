### Pickle

**Hintergrundwissen:**  
when ein programm beendet wird, wird der Arbeitsspeicher geleert,  
d.h. alle Objekte gehen nach der Programmbeendung verloren  
Pickle verhindert genau das, es dient als ein behälter und speichert die objekte auf die festplatte, ähnlich wie JSON

```python
# pickle in eine klasse methode einbauen, damit ein objekt trotz der beendung des programms nicht entsorgt wird

import pickle

class Student:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter
        
    def in_pickle_einpacken(self): # fachlich heißt einpacken 'serialisieren'
    
        behaelter = {'name':self.name, 'alter':self.alter} # zuerst die attribute bereitstellen
                  
        with open("beispiel_text_skript6", 'wb') as f: # weil die pickle-datei in binär geschrieben ist, ist die option 'b' nötig
            pickle.dump(behaelter, f) # .dump() wirft das zu speichernde objekt in pickle
        
    @classmethod 
    def aus_pickle_auspacken(cls): # fachlich heißt es 'deserialisieren'      
        with open("beispiel_text_skript6", 'rb') as f: 
            attribute = pickle.load(f) # .load() leert alle objekte in pickle aus 
        return cls (attribute["name"], attribute["alter"]) # die variale 'attribute' ist inhaltlich mit 'behaelter' identisch    
        
    def __str__(self):
        return f"der name ist {self.name}, das alter ist {self.alter}"
```

```python
a = Student("ken", 20)
```

```python
print(a)
```

```python
a.in_pickle_einpacken()
# "beispiel_text_skript6" ist nicht mehr auf jupyter notebook zu öffnen, d.h. die speicherung in pickle hat geklappt
```

```python
b = a.aus_pickle_auspacken()
```

```python
print(b)
```

### Classmethod  
**Vergleich zum Gelernten:**  
die methoden, die wir bis jetzt gelernt haben, sind instanzmethoden  
die sind aufrufbar, erst wenn ein objekt nach seiner erzeugung bereits existiert

**die anwendung**  
oft wird ein objekt erzeugt durch den aufruf der init-methode von intern, ohne dass wir es manuell erzeugen 

```python
class Auto:
    def __init__(self, marke, farbe):
        self.marke = marke
        self.farbe = farbe
        
    @classmethod
    def intern_erzeugen(cls, erzeugen_list):
        # oft wird das argument der cls-method erst nach verarbeitung zum erzeugen verwendet
        return cls(erzeugen_list[0], erzeugen_list[1])
```

```python
# normales erzeugen
a = Auto("mercedez", "schwarz")
```

```python
a
```

```python
# mit der classmethod ein objekt erzeugen

erzeugen_list = ["bmw", "weiß"]
b = Auto.intern_erzeugen(erzeugen_list) # direkt den namen der klasse und die klassmethode eingeben
```

```python
print(b.marke)
print(b.farbe)
```
