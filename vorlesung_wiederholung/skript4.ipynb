### Magische Methode in der Klasse

```python
# die magische methode __call__ macht ein objekt zu einer funktion

class Addtion:
    def __init__(self, summand):
        self.summand = summand
    def __call__(self, eingabe):
        return self.summand + eingabe
```

```python
a = Addtion(1)
b = 2
# a.__call__(b)
a(b)
```

```python
# die im wesentlichen mit __call__ gleiche Funktion

def addition(eingabe):
    summand=1
    return summand+eingabe
```

```python
b = 2
addition(b)
```

```python
# die methode __str__

class Student:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter
    def __str__(self):
       return f"der/die Studierende heißt {self.name} und ist {self.alter} jahre alt"
```

```python
a = Student('ken',20)
print(a)
```

```python
# attribut --> im vergleich zu einer methoden ist attribut ohne klammern aufzurufen
# attribut gilt quasi als eine variable. beide speichern, aber attribute gehören alle zu ihrem eigenen objekt
a.name
a.alter
```

### Zip

### Key in Sortierung & Lambda 

```python
# das "key" argument in .sort() oder sorted() 
# --> nach den regeln bzw. der funktion im "key" argument werden die elemente sortiert

# normalerweise wird dictionary durch nach seinem schlüssel sortiert
d = {"birne":5, "apfel":2, "dattel":1}
print(sorted(d))
```

```python
# durch key die regeln zur sortierung definieren
print(sorted(d, key=d.get)) # das argument von key erwartet nur den funktionsnamen, die klammern sind weggel
```

```python
# für key ist selbst definierte funktion auch möglich
def d_wert(x): # x ist die eingabe, beim sortieren eines dictionarys ist default der schlüssel
    return d[x]
print(sorted(d, key=d_wert))
```

```python
# lambda-funktion ist eine normale funktion halt ohne namen --> die funktion oben hat nur argument und return
sorted(d, key=lambda x:d[x])
```

```python
sorted(d, key=lambda x:-d[x])
```

```python
# lambda kann noch in anderen methoden verwendet werden, die ein iterierbares objekt erwartet, z.B. .max()
max(d, key=lambda x:d[x])
```

### Kompaktere Schreibweise

```python
# list comprehension --> eine abkürzung für for-schleife + list

wort = "halo"
buchstaben = []
for i in wort:
    buchstaben.append(i)
buchstaben
```

```python
buchstaben_list_comprehension = [i for i in wort] # an der stelle i ist das hinzuzufügende element
buchstaben_list_comprehension
```

```python
# generator expression ist ähnlich wie list comprehension

# erzeugung mit runden klammern
buchstaben_generator_expression = (i for i in wort)
buchstaben_generator_expression
```

```python
# erzeugung mit dem befehl yield, das kann nur in funktion erfolgen
def test_generator_expression():
    for i in wort:
        yield i
buchstaben_generator_expression1 = test_generator_expression()
buchstaben_generator_expression1
```

```python
# die unterschiede zwischen list comprehension und generator expression
# --> die elemente im list comprehension ist durch index zugreifbar, während das beim generator expression nicht möglich ist
# ----> list comprehension ersellt sofort die komplette liste im arbeitspeicher, 
# ----> während generator expression erst bei bedarf (for-schleife) die elemente berechnet 
print(buchstaben_list_comprehension[3])
```

```python
# print ist kein "bedarf" für generator expression, sondern for-schleife
print(buchstaben_generator_expression[3])
```

```python
# generator ist nicht wiederverwendbar, eine erstellung wird einmalig aufgebraucht
# list comprehension ist wiederverwendbar 
for i in buchstaben_generator_expression:
    print(i)
```

```python
# dictionary comprehension ist ähnlich wie list-comprehension
a = [1,2,3]
b = ["a","b","c"]
dictionary_comprehension = {i:k for k, i in zip(a,b)}
dictionary_comprehension
```
