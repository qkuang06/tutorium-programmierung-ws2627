# Linux

### Dateitypen

**Beim Ausfüren des Befehls 'ls -l' steht das erste Zeichen für den Dateityp**  
`-` : reguläre Datei  
`d`: Ordner

### Standard-Datenströme

**Jeder Prozess unter Linux öffnet automatisch drei Standard-Kommunikationskanäle**


`stdin` --> Standardeingabe  
`stdout` --> Standardausgabe  
`stderr` --> Standardfehlerausgabe

### Umleitungs- und Pipe-Operatoren

`<` Eingabeumleitung  
`>` Ausgabeumleitung (Überschreiben)  
*echo "hallo word" > text.txt*  
`>>` Ausgabeumleitung (Anhängen)  
*echo "hallo word" >> text.txt*  
`2>` Fehlerumleitung  
`2>&1` Zusammenführung von **stderr** und **stdout**  
`|` Pipe: Verbindet die stdout des linken Befehls direkt mit der stdin des rechten Befehls  

## cat (concatenate)

**cat liest Dateien ein und schreibt den Inhalt auf stdin also default direkt am terminal**
1. `cat text.txt` : die Datei lesen und den Inhalt im Terminal ausgeben
2. `cat text.txt > text1.txt` : die Datei lesen und den Inhalt in eine andere Datei reinschreiben
3. `cat < text.txt > text1.txt` : den Inhalt einer Datei bekommen und ihn in eine andere Datei reinschreiben

**der Befehl 2. und 3. funktioniert technisch gleich**

### Filter-Programme
**Definition: Eingabe + Verarbeitung + Ausgabe**  
`head` | `tail` | `wc (word count)` | `sort` | `grep (global regular expression print)` | `uniq (unique)` | `tr (translate)`

`head -n X`: standardmäßig die ersten zehn obersten Zeilen eines Texts ausgeben, X ist die Angabe der Anzahl der Zeilen

*head text.txt   
head -n 5 text.txt*

`tail -n X`: standardmäßig die letzen zehn untersten Zeilen eines Texts ausgeben, X ist die Angabe der Anzahl der Zeilen

*tail text.txt  
tail -n 5 text.txt*

`wc` zählt die Bestandteile je nach Optionen unterschiedlich
- `wc -l` : lines
- `wc -w` : wörter
- `wc -m` : zeichen'

`sort` : den Text zeilenweise alphabetisch sortieren
- `sort -n` : zeilenweise numerisch
- `sort -r` : zeilenweise alphabetisch absteigend Z->A

`grep` : die zeilen nach einem Muster durchsuchen, und die zeilen ausgeben, in denen das Muster vorkommt

*grep "an" text2.txt* --> findet alle zeilen, in denen "an" vorkommt

`uniq` : die doppelte zeilen, die direkt untereinander stehen, löschen bzw zu einer zeile reduzieren  
`uniq -c` : die anzahl der vorkommtnisse der gleichen zeile

*sort text.txt | uniq*

`tr` : zeichen ersetzen, ähnlich wie `re.sub()` oder `.replace()`

*echo "Hallo Linux" | tr "a-z" "A-Z"*
