### Aufgabe 3)

Das harmonische Mittel von n positiven reellen Zahlen ist definiert als die Anzahl der Zahlen geteilt durch die Summe ihrer Kehrwerte

```python
def harmonisches_mittel_brechnen():
    zahlen_liste = []
    while True:
        eingabe = input("geben Sie Zahlen ein, leere Zeile steht für Beenden: ")
        if eingabe == "":
            break
        eingabe = float(eingabe)
        zahlen_liste.append(eingabe)
    kehrwertsumme = 0
    for i in zahlen_liste:
        kehrwertsumme += 1 / i
    harmonisches_mittel = len(zahlen_liste) / kehrwertsumme
    return harmonisches_mittel
```

```python
harmonisches_mittel_brechnen()
```

### Aufgabe 4)

```python
def fibonacci():
    n = int(input("Geben Sie n ein: "))
    anfangsliste = [0, 1]

    for i in range(2, n + 1):
        naechste_zahl = anfangsliste[i - 2] + anfangsliste[i - 1]
        anfangsliste.append(naechste_zahl)

    print(f"Die {n}-te Fibonacci-Zahl ist: {anfangsliste[n]}")
```

### Aufgabe 3)
Die Definition des harmonischen Mittelwerts: Der Kehrwert des mittleren Kehrwerts.

```python
def harmonisches_mittel():
    zahl_liste = []
    while True:
        eingabe = input("geben Sie Zahlen ein, leere Zeile ist beenden: ")
        if eingabe == "":
            break
        eingabe = float(eingabe)
        zahl_liste.append(eingabe)
        
    if zahl_liste:
        kehrwertssumme = 0
        for i in zahl_liste:
            kehrwertssumme += 1 / i
        n = len(zahl_liste)
        harmonisches_mittel = n / kehrwertssumme
    return harmonisches_mittel
```

```python
print(harmonisches_mittel())
```
