Übung: Warenkorb-AnalyseThemen: Listen, Sets (Mengenlehre), for-Schleifen, f-Strings (Ausrichtung)

Schreiben Sie ein Programm, das die Warenkörbe von zwei Kunden vergleicht und übersichtlich ausgibt.

Aufgabenstellung:

1. Definieren Sie eine Funktion warenkorb_vergleich(korb1, korb2), die zwei Listen mit Strings (Artikelnamen) als Parameter entgegennimmt.

2. Wandeln Sie die Listen in Sets um, um Duplikate zu filtern.

3. Ermitteln Sie die Schnittmenge (Artikel, die in beiden Körben liegen) und die Vereinigungsmenge (alle Artikel aus beiden Körben zusammen).

4. Wandeln Sie die Vereinigungsmenge zurück in eine Liste und sortieren Sie diese alphabetisch.   

5. Geben Sie die sortierte Vereinigungsmenge mithilfe einer for-Schleife und f-Strings als formatierte Tabelle wie folgende aus. 

6. Die Artikelnamen sollen dabei linksbündig auf 15 Zeichen aufgefüllt werden (<15).   

```python
warenkorb1 = ["Apfel", "Banane", "Milch", "Banane"]
warenkorb2 = ["Milch", "Brot", "Apfel", "Käse"]

def warenkorb_vergleich(korb1:list, korb2:list):
    korb1 = set(korb1)
    korb2 = set(korb2)
    schnittmenge = korb1 & korb2
    vereinigung = korb1 | korb2
    vereinigung = list(vereinigung)
    vereinigung.sort()
    print(f"{'artikel':<15} | {'in beiden körben?':<15}")
    for i in vereinigung:
        print(f"{i:<15} | {'ja' if i in schnittmenge else 'nein'}")
    
    
    
```

```python
warenkorb_vergleich(warenkorb1, warenkorb2)
```

Übung: Passwort-SicherheitsprüferThemen: Funktionen, Strings (Methoden), if/else (Einzeiler), len()

Schreiben Sie ein Programm, welches ein vom Benutzer eingegebenes Passwort auf grundlegende Sicherheitskriterien prüft.

Aufgabenstellung:

1. Schreiben Sie eine Funktion passwort_check(passwort), die einen String als Argument erhält und einen Boolean (True oder False) zurückgibt.   

2. Ein Passwort gilt als sicher, wenn es alle folgenden Kriterien erfüllt:
    1) Es ist mindestens 8 Zeichen lang.   
    2) Es enthält mindestens einen Buchstaben (Tipp: Prüfen Sie die einzelnen Zeichen mit .isalpha()).   
    3) Es enthält mindestens eine Zahl (Tipp: Prüfen Sie die einzelnen Zeichen mit .isdigit()).   

3. Nutzen Sie eine for-Schleife, um über die Zeichen des Passwort-Strings zu iterieren und Variablen hochzuzählen oder Booleans zu setzen, falls Buchstaben oder Zahlen gefunden werden.   

4. Rufen Sie die Funktion am Ende des Skripts mit einem Beispielpasswort auf.

5. Nutzen Sie eine einzeilige if/else-Bedingung (Ternary Operator), um das Ergebnis der Funktion auszuwerten und entweder "Passwort sicher" oder "Passwort unsicher" auf der Konsole auszugeben.   

```python
def aufrufen():
    passwort = input('geben Sie Ihr passwort ein')
    passwort_check(passwort)
    
def passwort_check(passwort:str) -> bool:
    min_ein_buchstabe = False
    min_eine_zahl = False

    for i in passwort:
        if i.isalpha():
            min_ein_buchstabe = True
        if i.isdigit():
            min_eine_zahl = True

    laenge_groesser = False
    if len(passwort) >= 8:
        laenge_groesser = True

    if laenge_groesser and (min_ein_buchstabe and min_eine_zahl):
        print('das passwort ist sicher')
    else:
        print('das passwort ist unsicher')
```

```python
aufrufen()
```
