###  Aufgabe 1 — Bestellungen einlesen und säubern
Schreibe eine Funktion bestellungen_einlesen(dateiname), die:

1. Die Datei öffnet.
2. Jede Zeile in Kundenname und Pizza-Liste aufteilt (Trenner: |).
3. Alle unnötigen Leerzeichen entfernt und Namen & Pizzen einheitlich klein- bzw. großgeschrieben (Name: Kapitalisiert, Pizzen: klein).
4. Ein Dictionary zurückgibt, bei dem die Schlüssel die Kundennamen sind und die Werte Listen der bestellten Pizzen.

```python
def bestellungen_einlesen(dateiname):
    with open(dateiname, "r", encoding='utf-8') as f:
        kunde_pizza = {}
        for zeile in f:
            
            # 2. Jede Zeile in Kundenname und Pizza-Liste aufteilen
            kundenname = zeile.split('|')[0].strip().upper()
            pizzas = zeile.split('|')[1] 
            print(kundenname)
            print(pizzas)

            # 3. Alle unnötigen Leerzeichen entfernt und Namen & Pizzen einheitlich klein- bzw. großgeschrieben
            pizza_liste = pizzas.split(',')
            pizza_liste = [pizza.strip().lower() for pizza in pizza_liste]

            # 4. Ein Dictionary zurückgiben
            kunde_pizza[kundenname] = pizza_liste
    return kunde_pizza
```

### Aufgabe2 — Statistik-Report erzeugen  
Schreibe eine Funktion statistik(bestellungen), die aus dem Dictionary aus Aufgabe 1:

1. Die Gesamtzahl der verkauften Pizzen berechnet.
2. Mit collections.Counter ermittelt, wie oft jede Pizza-Sorte bestellt wurde.
3. Die Top 3 Pizzen sortiert nach Häufigkeit ermittelt — verwende sorted(...) mit einer Lambda-Funktion als key.
4. Den Lieblingskunden (der die meisten Pizzen bestellt hat) herausfindet

```python
from collections import Counter
from itertools import chain
def statistik(bestellungen:dict):
    
    # 1. Gesamtzahl
    gesamtzahl = sum(len(pizzas_liste) for pizzas_liste in bestellungen.values())
    print(gesamtzahl, '\n')

    
    # 2. Wie oft jede Pizza-Sorte bestellt wurde
    all_pizzas_liste = list(chain(*bestellungen.values()))
    print(all_pizzas_liste, '\n')
    pizza_sorte_anzahl = Counter(all_pizzas_liste)
    print(pizza_sorte_anzahl, '\n') # zufällig absteigend!! 


    # 3. Die Top 3 Pizzen
    sort_pizza_sorte_anzahl = sorted(pizza_sorte_anzahl, key=lambda x:-pizza_sorte_anzahl[x])
    print(sort_pizza_sorte_anzahl, '\n')
    
    
    # 4. Lieblingskunden
    pizza_anzahl = 0
    lieblingskunden = []
    for kunde, pizzas in bestellungen.items():
        if len(pizzas) < pizza_anzahl:
            continue
        elif len(pizzas) == pizza_anzahl:
            lieblingskunden.append(kunde)
        else:
            pizza_anzahl = len(pizzas)
            lieblingskunden.clear()
            lieblingskunden.append(kunde)
    print(lieblingskunden)

    # alternativ: max, aber max gibt nur ein element zurück, auch wenn Daniel und Eva genau so viel bestellt haben
    #lieblingskunden = max(bestellungen, key=lambda x:len(bestellungen[x])) 
    #print(lieblingskunden)
```

### Aufgabe 3 — Klasse Kunde mit Rabatt-System  
Erstelle eine Klasse **"Kunde"** mit folgenden Eigenschaften:
1. Attribute: name, bestellungen (Liste), stammkunde (bool, Default False)
2. Konstruktor `__init__` mit Default-Argumenten für bestellungen und stammkunde.
3. Methode pizza_bestellen(pizza), die eine Pizza zur Bestellliste hinzufügt.
4. Methode rechnung(), die den Gesamtpreis berechnet. Jede Pizza kostet 8,50 €. Stammkunden bekommen 10 % Rabatt. Ab 5 Pizzen zusätzlich 2 € Mengenrabatt.
5. Magische Methode `__str__`, die eine schöne Zusammenfassung zurückgibt, z. B. : Kunde Eva — 4 Pizzen bestellt — Rechnung: 30.60 €

```python
class Kunde:
    def __init__(self, name, bestellungen, stammkunde=False):
        self.name = name
        self.bestellungen = []
        self.stammkunde = stammkunde
    def pizza_bestellen(self, pizza):
        self.bestellungen.append(pizza)
    def rechnung(self):
        preis = len(self.bestellungen)*8.5
        if self.stammkunde:
            preis = preis*(1-0.1)
        if len(self.bestellungen)>=5:
            preis -= 2
        return preis
    def __str__(self):
        return f'Kunde {self.name} — {len(self.bestellungen)} Pizzen bestellt — {self.rechnung()}: 30.60 €'
```

### Teil B — Kunden-Ähnlichkeit (Set-Operationen)
Schreibe eine Funktion ähnlichkeit(bestellungen), die für jedes Kundenpaar zurückgibt:

Wie viele gemeinsame Pizza-Sorten sie haben (Schnittmenge &)  
Wie viele verschiedene Pizza-Sorten sie zusammen probieren (Vereinigung |)  
Rückgabe: Dictionary mit einem Tupel (kunde1, kunde2) als Schlüssel und einem Tupel (gemeinsam, gesamt) als Wert.

Wichtig: Paare dürfen nicht doppelt auftauchen — (Anna, Ben) und (Ben, Anna) sind dasselbe. Sortiere die Namen alphabetisch, bevor du sie als Schlüssel nutzt.

```python
def ähnlichkeit(bestellungen):
    kunde_set = {name: set(pizzen) for name, pizzen in bestellungen.items()}
    print(kunde_set, "\n") 
    ergebnis = {}
    kunden = list(kunde_set.keys())
    
    for i in range(len(kunden)):
        for j in range(i + 1, len(kunden)):
            k1 = kunden[i]
            k2 = kunden[j]
            
            # alphabetisch sortieren für einen eindeutigen schlüssel
            paar = tuple(sorted([k1, k2])) # der schlüssel muss unveränderbar sein
            
            gemeinsam = len(kunde_set[k1] & kunde_set[k2])
            gesamt = len(kunde_set[k1] | kunde_set[k2])
            
            ergebnis[paar] = (gemeinsam, gesamt)
    
    return ergebnis
```
