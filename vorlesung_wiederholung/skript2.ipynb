x and y:
falls x falsch ist, egal ob y wahr ist, ist die Aussage falsch. —> es gibt einfach x aus
falls x wahr ist, hängt es von dem element y ab, ob die aussage wahr oder falsch ist. —> es gibt y aus

x or y:
falls x falsch ist, hängt es von dem element y ab, ob die aussage wahr oder falsch ist —> es gibt y aus
falls x wahr ist, egal ob y wahr oder falsch ist, ist die aussage wahr —> es gibt x aus

```python
a b
a and b
a or b
```

### if-Bedingung

```python
# wenn es zwei situationen sind, schreiben wir oft if & else
a = 3
if a > 0:
    print('a ist größer als 0')
else:
    print('a ist kleiner gleich 0')
```

```python
# einzeiler
print('a ist größer als 0' if a > 0 else 'a ist kleiner gleich 0')
```

```python
# wenn es mehrerer situationen sind, schreiben wir oft if & elif & else
a = 0
if a > 0:
    print('a ist größer als 0')
elif a == 0:
    print('a ist gleich 0')
else:
    print('a ist kleiner als 0')
```

```python
# einzeiler
print('a ist größer als 0' if a > 0 else 'a ist gleich 0' if a == 0 else 'a ist kleiner gleich 0' )
```
